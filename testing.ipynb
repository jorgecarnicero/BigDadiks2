```python
import os
import boto3
from botocore.exceptions import ClientError

# 1) Pones credenciales en el entorno (aquí "solo código")
os.environ["AWS_ACCESS_KEY_ID"] = "TU_ACCESS_KEY_ID"
os.environ["AWS_SECRET_ACCESS_KEY"] = "TU_SECRET_ACCESS_KEY"
# Si usas credenciales temporales (por ejemplo, STS), añade:
# os.environ["AWS_SESSION_TOKEN"] = "TU_SESSION_TOKEN"

REGION = "eu-south-2"
BUCKET = "mi-bucket-unico-123456789"  # debe ser globalmente único

s3 = boto3.client("s3", region_name=REGION)

try:
    # En regiones distintas de us-east-1 debes pasar CreateBucketConfiguration
    s3.create_bucket(
        Bucket=BUCKET,
        CreateBucketConfiguration={"LocationConstraint": REGION},
    )
    print("Bucket creado:", BUCKET)

    # Verificar (HEAD)
    s3.head_bucket(Bucket=BUCKET)
    print("Verificado OK:", BUCKET)

except ClientError as e:
    print("Error:", e.response["Error"]["Code"], "-", e.response["Error"].get("Message", ""))
```
